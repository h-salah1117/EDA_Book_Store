```python
import numpy as np
import matplotlib.pyplot as plt
from dash import Dash, dcc, html, Input, Output
import pandas as pd
import plotly.express as px
import seaborn as sns


```

```python
df = pd.read_csv('Books_Data_Clean.csv')
df.describe()
df.info()
df.head()

```

```python
print((df['Publishing Year'] > 1900).sum())
df.describe()
```

```python
df['Publishing Year'] = df['Publishing Year'].apply(lambda x: x if x > 1900 else np.nan)
df.dropna(subset=['Publishing Year', 'Book Name'], inplace=True)
df.describe()
```

```python
Publisher_Revenue_Distribution = px.histogram(df, x='publisher revenue', 
                   title='Publisher Revenue Distribution',
                   nbins=50)
Publisher_Revenue_Distribution.show()
```

```python
import plotly.express as px

Book_Ratings_vs_Gross_Sales = px.scatter(
    df,
    x='Book_average_rating',
    y='gross sales',
    size='Book_ratings_count',
    color='genre',
    hover_name='Book Name',
    title='Book Ratings vs Gross Sales',
    labels={
        'Book_average_rating': 'Average Rating',
        'gross sales': 'Gross Sales (Revenue)',
        'Book_ratings_count': 'Number of Ratings',
        'genre': 'Genre'
    },
    template='plotly_white',
    size_max=40,      
    opacity=0.7        
)

Book_Ratings_vs_Gross_Sales.show()
```

```python
top_authors_rating = (
    df.groupby('Author')['Book_average_rating'].mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
Average_Rating_by_Author =px.bar(top_authors_rating, x='Author', y='Book_average_rating', title='Average Rating by Author',color='Book_average_rating',
             color_continuous_scale='viridis')
Average_Rating_by_Author.update_yaxes(range=[3.5, 4.5]) 
Average_Rating_by_Author.show()
```

```python
top_authors_gross_sales = (
    df.groupby('Author')['gross sales'].sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
fig =px.bar(top_authors_gross_sales, x='Author', y='gross sales', title='Total Gross Sales by Author',color='gross sales',
             color_continuous_scale='viridis')
fig.show()
```

```python
Box_Plot_of_Book_Ratings_by_Genre = px.box(df, x='genre', y='Book_average_rating', color='genre',
             title='Box Plot of Book Ratings by Genre')
Box_Plot_of_Book_Ratings_by_Genre.show()
```

```python
mean_gross=df.groupby('genre')['gross sales'].mean().reset_index()
mean_gross=mean_gross.sort_values(by='gross sales',ascending=False)
Total_Gross_Sales_by_Genre = px.bar(
    mean_gross, x='genre', y='gross sales', color='genre',
    title='Total Gross Sales by Genre'
)
Total_Gross_Sales_by_Genre.show()
```

```python
df['Publishing Year'] = pd.to_numeric(df['Publishing Year'], errors='coerce')

# Aggregate by year and genre
yearly = (
    df.groupby(['Publishing Year', 'genre'], as_index=False)
      .agg({
          'Book_average_rating': 'mean',
          'gross sales': 'sum'
      })
)

# --- Choose the metric to visualize ---
metric = "gross sales"  # or "Book_average_rating"

# --- Create the interactive line chart with a slider ---
fig = px.line(
    yearly,
    x="Publishing Year",
    y=metric,
    color="genre",
    markers=True,
    title=f"Book {metric.title()} Trends Over Time by Genre",
    labels={
        "Publishing Year": "Publishing Year",
        metric: metric.replace("_", " ").title(),
        "genre": "Genre"
    },
)

# --- Customize layout with a year range slider ---
fig.update_layout(
    title_font_size=20,
    legend_title="Genre",
    hovermode="x unified",
    template="plotly_dark",  # use "plotly_white" if you prefer light mode
    width=1000,
    height=600,
    xaxis=dict(
        rangeslider=dict(
            visible=True,           # 🔹 enables the slider
            bgcolor="rgba(100,100,100,0.1)"
        ),
        rangeselector=dict(        # 🔹 optional buttons for quick range selection
            buttons=list([
                dict(count=5, label="Last 5 yrs", step="year", stepmode="backward"),
                dict(count=10, label="Last 10 yrs", step="year", stepmode="backward"),
                dict(step="all", label="All Years")
            ])
        )
    )
)

fig.show()
```
